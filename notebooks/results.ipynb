```python
import pandas as pd
import numpy as np
```

```python
log = pd.read_csv("results/train_log.csv")
log["reward_smooth"] = log["mean_reward"].rolling(50, min_periods=1).mean()
ax = log.plot(x="step", y=["mean_reward", "reward_smooth"], alpha=0.6)
ax.set_ylabel("mean ΔSI-SDR (dB)")
```

```python
df = pd.read_csv("results/test_results.csv")
summary = (df.groupby("method")["delta_si_sdr"]
             .agg(["mean", "median", "std", "count"]).round(2))
summary
```
