```python
import librosa
from IPython.display import HTML
import matplotlib.pyplot as plt
from IPython.display import Audio
import numpy as np
import noisereduce as nr
```

```python
filename = librosa.example("trumpet")
y, sr = librosa.load(filename)

HTML(librosa.util.example_info("trumpet", html=True))
```

```python
# Construct a figure and axes object:
fig, ax = plt.subplots()

# Plot on the axes:
librosa.display.waveshow(y, sr=sr, ax=ax)

# Set the title for the figure
ax.set(title="Trumpet example")
```

```python
Audio(data=y, rate=sr)
```

```python
stft = librosa.stft(y)
print('stft shape:', stft.shape)
fig, ax = plt.subplots()
img = librosa.display.specshow(stft, x_axis="time", y_axis="hz", vscale="dB")
librosa.display.colorbar_db(img)
```

```python
reduced_noise = nr.reduce_noise(y = y, sr=sr, n_std_thresh_stationary=1.5,stationary=True)
```

```python
# Construct a figure and axes object:
fig, ax = plt.subplots()

# Plot on the axes:
librosa.display.waveshow(y=reduced_noise, sr=sr, ax=ax)

# Set the title for the figure
ax.set(title="Trumpet example")
```

```python
reduced_stft = librosa.stft(reduced_noise)
print('stft shape:', reduced_stft.shape)
fig, ax = plt.subplots()
img = librosa.display.specshow(reduced_stft, x_axis="time", y_axis="hz", vscale="dB")
librosa.display.colorbar_db(img)
```

```python
Audio(data=reduced_noise, rate=sr)
```
